```python
match_data_dir = os.listdir(match_csv_path)
match_data_dir
```

```python
def process_match_sheet(full_df, match_csv_path, sheet_name):
    match_df = pd.read_csv(match_csv_path + sheet_name, header = None)
    match_df.drop(29, axis=1, inplace=True)
    match_df.columns = full_df.columns
    full_df = pd.concat([full_df, match_df], ignore_index = True)
    return full_df
```

```python
for file_name in match_data_dir:
    full_df = process_match_sheet(full_df, match_csv_path, file_name)
    shutil.move(match_csv_path + file_name, processed_csv_path + file_name)
full_df
```

```python
full_df['AutoAccuracyNum'] = full_df['AutoAccuracy'].str[-3:-1].astype(float)/100
full_df['TeleopAccuracyNum'] = full_df['TeleopAccuracy'].str[-3:-1].astype(float)/100
columns_to_encode = ['AutoHang','HangStatus','AccuracyPos','OverBump','UnderTrench','PlayedDefense','CollectedFuel','PassedFuel','DidNothing','Other','Defense','StopReason']
full_df = pd.get_dummies(full_df, columns=columns_to_encode)
print("\nEncoded DataFrame:")
full_df
```
